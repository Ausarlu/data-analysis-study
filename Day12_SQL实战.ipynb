```python
import yfinance as yf
import pandas as pd

def get_stock_data(ticker, start_date, end_date):
    data = yf.download(ticker, start=start_date, end=end_date)
    data.columns = data.columns.get_level_values(0)
    return data

tickers = ['AAPL', 'MSFT', '600519.SS', '000001.SZ']
all_data = {}

for ticker in tickers:
    all_data[ticker] = get_stock_data(ticker, '2025-01-01', '2026-01-01')
```

```python
import sqlite3 
```

```python
conn = sqlite3.connect('stock_data.db')
```

```python
maotai_data = all_data['600519.SS']
maotai_data.to_sql('maotai',conn,if_exists='replace')
```

```python
for ticker, data in all_data.items():
    table_name = ticker.replace('.', '_')
    data.to_sql(table_name,conn,if_exists='replace')
```

```python
tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn)
print(tables)
```

```python
result = pd.read_sql("select * from maotai limit 5", conn)
print(result)
```

```python
result2 = pd.read_sql("SELECT * FROM AAPL WHERE Close > 200 ORDER BY Close DESC " , conn)
print(result2)
```

```python
result3 = pd.read_sql("select max(High), min(Low) from AAPl where Date like '2025-12%' ",conn)
print(result3)
```
